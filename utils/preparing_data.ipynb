### Prepare images data

```python
import os, cv2
from procces_library import match_histograms
from metadata_gps import write_exif

input_dir= "test/"                ## images file only
output_dir= "ready_to_use/"
img_ref="images_reference/LEC.jpg"
for filename in os.listdir(input_dir):
    img_targ=input_dir+filename
    img_out=output_dir+filename
    img_ready=match_histograms(img_targ, img_ref)
    cv2.imwrite(img_out, img_ready)
    write_exif(img_targ, img_out)

```
